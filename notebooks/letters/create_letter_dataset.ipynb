```python
import requests
import os
import json
from utils.use_textract_utils import parse_local_pdfs_with_textract

download_dir = "/Users/melih.gorgulu/Desktop/Projects/aftercourt_automation/assets/letter_on_16june/pdfs"
cache_dir = "/Users/melih.gorgulu/Desktop/Projects/aftercourt_automation/assets/letter_on_16june"
cache_file = os.path.join(cache_dir, "_download_cache.json")
os.makedirs(download_dir, exist_ok=True)

# Load cache
if os.path.exists(cache_file):
    with open(cache_file, "r") as f:
        cache = json.load(f)
else:
    cache = {"downloaded": {}}  # attachment_id -> file_path

print(f"[Download Cache] Loaded {len(cache['downloaded'])} cached entries, {len(new_letter_df)} total to process")

# Download files using token URLs
local_paths = []
cache_hits = 0
for _, row in new_letter_df.iterrows():
    attachment_id = str(row['attachment_id'])
    url = row['attachment_url']
    ext = os.path.splitext(url.split("name=")[-1])[-1] if "name=" in url else ".pdf"
    file_path = os.path.join(download_dir, f"{attachment_id}{ext}")

    # Skip if already in cache
    if attachment_id in cache["downloaded"]:
        local_paths.append(cache["downloaded"][attachment_id])
        cache_hits += 1
        continue

    if not os.path.exists(file_path):
        resp = requests.get(url)
        if resp.status_code == 200:
            with open(file_path, "wb") as f:
                f.write(resp.content)
            local_paths.append(file_path)
            cache["downloaded"][attachment_id] = file_path
            # Save cache after each successful download
            with open(cache_file, "w") as f:
                json.dump(cache, f)
        else:
            print(f"Failed to download {attachment_id}: {resp.status_code}")
    else:
        local_paths.append(file_path)
        cache["downloaded"][attachment_id] = file_path

# Final cache save
with open(cache_file, "w") as f:
    json.dump(cache, f)

print(f"Cache hits: {cache_hits}, newly downloaded: {len(local_paths) - cache_hits}, total: {len(local_paths)}")
```

```python
textract_cache_file = os.path.join(download_dir, "_textract_cache.json")
if os.path.exists(textract_cache_file):
    with open(textract_cache_file, "r") as f:
        textract_cache = json.load(f)
    print(f"Textract cache has {len(textract_cache)} entries")
else:
    print("No textract cache found")
```

```python
# Parse downloaded PDFs with Textract (with caching)
textract_cache_file = os.path.join(download_dir, "_textract_cache.json")

if os.path.exists(textract_cache_file):
    with open(textract_cache_file, "r") as f:
        textract_cache = json.load(f)
else:
    textract_cache = {}  # s3_object_key -> extracted_text

print(f"[Textract Cache] Loaded {len(textract_cache)} cached entries, {len(local_paths)} total to process")

# Filter out already-parsed files (only PDFs)
paths_to_parse = [p for p in local_paths if p.lower().endswith(".pdf") and f"ocr_source_files/letters/{os.path.basename(p)}" not in textract_cache]
print(f"[Textract Cache] Cache hits: {len(local_paths) - len(paths_to_parse)}, to parse: {len(paths_to_parse)}")

# Process in batches of 100, caching results after each batch
BATCH_SIZE = 100
for batch_start in range(0, len(paths_to_parse), BATCH_SIZE):
    batch = paths_to_parse[batch_start:batch_start + BATCH_SIZE]
    batch_num = batch_start // BATCH_SIZE + 1
    total_batches = (len(paths_to_parse) + BATCH_SIZE - 1) // BATCH_SIZE
    print(f"[Textract] Processing batch {batch_num}/{total_batches} ({len(batch)} files)")

    new_results = parse_local_pdfs_with_textract(
        local_pdf_paths=batch,
        s3_object_key_base="ocr_source_files/letters",
        use_page_markers=True,
    )
    # Update and persist cache after each batch
    textract_cache.update(new_results)
    with open(textract_cache_file, "w") as f:
        json.dump(textract_cache, f)
    print(f"[Textract] Cached batch {batch_num}/{total_batches}, total cached: {len(textract_cache)}")

# Use full cached results
results = textract_cache
```
