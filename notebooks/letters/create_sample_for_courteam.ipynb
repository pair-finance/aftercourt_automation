```python
dfs = [sample_da_no_invoice]
names = ["sample_da_no_invoice"]
```

```python
import sys
import boto3
import os
sys.path.append('/Users/melih.gorgulu/Desktop/Projects/aftercourt_automation')
from utils.prod_utils import download_pdf_from_s3

session = boto3.Session(
    region_name="eu-central-1",
    profile_name="739275445236_DataScienceUser",
)
s3 = session.client("s3")

S3_BUCKET = "pair-email-classification"

for df, name in zip(dfs, names):
    group_dir = os.path.join(download_path, name)
    os.makedirs(group_dir, exist_ok=True)
    print(f"\n--- Downloading {len(df)} files into {group_dir} ---")
    for _, row in df.iterrows():
        try:
            download_pdf_from_s3(
                s3=s3,
                bucket_name=S3_BUCKET,
                object_key=row["s3_object_key"],
                download_dir=group_dir,
            )
        except Exception as e:
            print(f"  ❌ Failed for {row['s3_object_key']}: {e}")
```
