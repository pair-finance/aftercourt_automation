```python
# fetch attachment ids

def get_llm_reasoning_by_attachment_id(attachment_ids, analytics_db, graph_db):
    attachment_ids_str = ",".join([f"'{id}'" for id in attachment_ids])
    query = f"""
    SELECT ticket_uuid
    FROM llm_attachments
    WHERE attachment_id IN ({attachment_ids_str})
    """
    ticket_uuids = analytics_db.sql_to_df(query)
    
    query = f"""
    SELECT *
    FROM public.gpu_tasks
    WHERE ticket_uuid IN ({','.join([f"'{uuid}'" for uuid in ticket_uuids['ticket_uuid'].tolist()])})
    """
    gpu_tasks = graph_db.sql_to_df(query)
    return gpu_tasks
```
