```python
#### New Tag: Geschäftsnummer

# Updated regex to match patterns like: 5 am 3951/23, 5a m 905/24, 5 am 4087/22, 7m 797/23

regex_geschaeftsnummer = r"(?i)\b(?:\d+\s*a?\s*m\s*\d+(?:\/\d+)?|m\s+\d+\/\d+)\b"
import re
```

```python
data['geschaeftsnummer'] = data['cleaned_text'].apply(lambda x: re.findall(regex_geschaeftsnummer, x))
data['has_geschaeftsnummer'] = data['geschaeftsnummer'].apply(lambda x: len(x) > 0)
```

```python
has_geschaeftsnummer_NO_PFUB = data[(data['has_geschaeftsnummer'] == True) & (data['is_pfub'] == False)]
NOT_has_geschaeftsnummer_PFUB = data[(data['has_geschaeftsnummer'] == False) & (data['is_pfub'] == True)]
```
