### Further Preprocessing: Spacy Tokenization

```python
class CustomTokenizer(ClassificationSpacyLemmaTokenizer):
    def __call__(self, text):
        doc = self.nlp(text)
        tokens_cleared = [
            token.lemma_.lower()
            for token in doc
            if (
                (not token.is_stop and not token.is_punct and not token.like_num or token == "m")
            ) and (
                token.lemma_ != "\n" and token.lemma_ != "\n\n" and 
                token.lemma_ != " " and token.lemma_ != "" and 
                len(token.lemma_) < 45
            ) and
            (not token.lemma_.startswith("<") and not token.lemma_.endswith(">"))  # Exclude special tokens
        ]
        return tokens_cleared

tokenizer = CustomTokenizer()
```

```python
data['tokenized'] = data['text_w_tags'].apply(lambda x: tokenizer(x))
```

### Lets check what are the common tokens in pfub erlass documents

```python
pfub_erlass_voc = {}
total_documents = len(pfub_erlass)
for tokens in pfub_erlass['tokenized']:
    for token in tokens:
        if token in pfub_erlass_voc:
            pfub_erlass_voc[token] += 1
        else:
            pfub_erlass_voc[token] = 1

for token in pfub_erlass_voc:
    pfub_erlass_voc[token] = round(pfub_erlass_voc[token] / total_documents, 2)  # normalize by number of documents
            
pfub_erlass_voc = dict(sorted(pfub_erlass_voc.items(), key=lambda item: item[1], reverse=True))
```

```python
pfub_erlass_voc
```

```python
frequent_pfub_erlass_tokens = {k: v for k, v in pfub_erlass_voc.items() if v >= 0.1}
frequent_pfub_erlass_tokens
```

```python
not_pfub_erlass = data[data['is_pfub'] == 0]

# given frequent tokens in pfub erlass, check their frequencies in not pfub erlass
frequent_pfub_erlass_tokens_in_not_pfub = {}
for token in frequent_pfub_erlass_tokens:
    count = not_pfub_erlass['tokenized'].apply(lambda x: x.count(token)).sum()
    frequent_pfub_erlass_tokens_in_not_pfub[token] = round(count/not_pfub_erlass.shape[0], 2)
frequent_pfub_erlass_tokens_in_not_pfub
```

```python
differences = {}
for token in frequent_pfub_erlass_tokens:
    diff = frequent_pfub_erlass_tokens[token] - frequent_pfub_erlass_tokens_in_not_pfub.get(token, 0)
    differences[token] = round(diff, 2)
```

```python
significant_differences = {k: v for k, v in differences.items() if v >= 0.1}
significant_differences
```

```python
# plot first 20 significant differences as a bar chart
significant_differences_sorted = dict(sorted(significant_differences.items(), key=lambda item: item[1], reverse=True))
tokens = list(significant_differences_sorted.keys())[:30]
values = [significant_differences_sorted[token] for token in tokens]
plt.figure(figsize=(12, 6))
plt.bar(tokens, values)
plt.xlabel('Tokens')
plt.ylabel('Difference in Frequency')
plt.title('Top 20 Tokens with Significant Frequency Differences (pfub-erlass vs not pfub-erlass)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```

```python
# take top 10 token, scatter plot, x axis: token, y axis: frequency. group by document_type
plot_data = {}

for token in list(significant_differences_sorted.keys())[:10]:
    plot_data[token] = {}
    for doc_type in data['document_type'].unique():
        count = data[data['document_type'] == doc_type]['tokenized'].apply(lambda x: x.count(token)).sum()
        plot_data[token][doc_type] = round(count / data[data['document_type'] == doc_type].shape[0], 2)
```

```python
# plot plot data
for token in plot_data:
    plt.figure(figsize=(8, 5))
    doc_types = list(plot_data[token].keys())
    frequencies = list(plot_data[token].values())
    # if doc_type in ['approved_attachment_and_transfer_order', 'approved_seizure'], use x to plot
    for i, doc_type in enumerate(doc_types):
        if doc_type in ['approved_attachment_and_transfer_order', 'approved_seizure']:
            plt.scatter(doc_type, frequencies[i], alpha=0.7, s=100, marker='X')
        else:
            plt.scatter(doc_type, frequencies[i], alpha=0.7, s=100)
    plt.xlabel('Document Type')
    plt.ylabel('Frequency')
    plt.title(f'AVG Token Frequency by Document Type for Token: <{token}>')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.legend()
    plt.show()
    
    # NOTE: some keywords are similar in approved_attachment_and_transfer_order and attachment_and_transfer_order. tHIS MAY CAUSE CONFUSION
    # If we sure that we wont have attachment_and_transfer_order input (since it is not required in OCR), we can ignore this data source.
```

### Aplly TF-IDF TO see important tokens for PFUB ERLASS

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(tokenizer=CustomTokenizer(),
                             max_features=300, # it was 200 before
                             ngram_range=(1,1), #it was (1,3) before,
                             norm='l2',
                             min_df=2,
                             max_df=0.95,
                             stop_words=None,
                             lowercase=False)
```

```python
vectorizer.fit(data['text_w_tags'])
```

```python
vectorized_corpus = vectorizer.transform(data['text_w_tags'])
```

```python
vectorizer.get_feature_names_out()
```

```python
feature_names = vectorizer.get_feature_names_out()
pfub_indexes = data[data['is_pfub'] == 1].index.tolist()
pfub_vectors = vectorized_corpus[pfub_indexes]
pfub_vectors.toarray()
pfub_vectors_df = pd.DataFrame(pfub_vectors.toarray(), columns=feature_names)
pfub_vectors_df
```

```python
mean_pfub_vector = pfub_vectors_df.mean().sort_values(ascending=False)
mean_pfub_vector
```

```python
top_30_pfub_tfidf = mean_pfub_vector.head(30)
```

```python
top_30_pfub_tfidf.index
```

```python
# For top 30 tfidf tokens in pfub erlass, check their frequencies different document types
token_frequencies_by_doc_type = {}
for token in top_30_pfub_tfidf.index:
    token_frequencies_by_doc_type[token] = {}
    for doc_type in data['document_type'].unique():
        count = data[data['document_type'] == doc_type]['tokenized'].apply(lambda x: x.count(token)).sum()
        token_frequencies_by_doc_type[token][doc_type] = round(count / data[data['document_type'] == doc_type].shape[0], 2)

# now splot it as a heatmap
!pip install seaborn
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 8))
token_list = list(token_frequencies_by_doc_type.keys())
doc_type_list = list(data['document_type'].unique())
heatmap_data = np.zeros((len(token_list), len(doc_type_list)))
for i, token in enumerate(token_list):
    for j, doc_type in enumerate(doc_type_list):
        heatmap_data[i][j] = token_frequencies_by_doc_type[token][doc_type]

sns.heatmap(heatmap_data, xticklabels=doc_type_list, yticklabels=token_list, cmap="YlGnBu")
plt.xlabel("Document Type")
plt.ylabel("Token")
plt.title("Token Frequencies by Document Type for Top 30 PFUB TF-IDF Tokens")
plt.show()
```

```python
# now plot heatmap for each document type, using tf idf values of the top 30 pfub tokens, given the document type. Heatmap data should come from tf idf values.
# use one plot, x is token, y is document type, value is mean tf idf value of that token in that document type

plt.figure(figsize=(12, 8))
token_list = top_30_pfub_tfidf.index.tolist()
doc_type_list = list(data['document_type'].unique())
heatmap_data = np.zeros((len(doc_type_list), len(token_list)))
for i, doc_type in enumerate(doc_type_list):
    doc_indexes = data[data['document_type'] == doc_type].index.tolist()
    doc_vectors = vectorized_corpus[doc_indexes]
    doc_vectors_df = pd.DataFrame(doc_vectors.toarray(), columns=feature_names)
    mean_doc_vector = doc_vectors_df.mean()
    for j, token in enumerate(token_list):
        heatmap_data[i][j] = mean_doc_vector[token]
sns.heatmap(heatmap_data, xticklabels=token_list, yticklabels=doc_type_list, cmap="YlGnBu")
plt.xlabel("Token")
plt.ylabel("Document Type")
plt.title("Mean TF-IDF Values of Top 30 PFUB Tokens by Document Type")
plt.show()
```
