```python
# # save the trained model
# save_path = os.path.join(notebook_dir, "after-court/models/classification/28-01-2026_binary_ladung_rf_classifier_v1.dill")
# with open(save_path, "wb") as f:
#     dill.dump(rf_classifier, f)
```

```python
ths = [0.1,0.5,0.6,0.7,0.8,0.90, 0.91, 0.92, 0.93, 0.95, 0.96, 0.97, 0.98, 0.99]

threshold_results = []

for th in ths:
    y_pred_label = y_pred_probs >= th
    
    # Calculate confusion matrix components
    tp = ((y_pred_label == 1) & (y_test == 1)).sum()
    fp = ((y_pred_label == 1) & (y_test == 0)).sum()
    tn = ((y_pred_label == 0) & (y_test == 0)).sum()
    fn = ((y_pred_label == 0) & (y_test == 1)).sum()
    
    # Calculate metrics
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    accuracy = (tp + tn) / (tp + tn + fp + fn)
    
    threshold_results.append({
        'threshold': th,
        'tp': tp,
        'fp': fp,
        'tn': tn,
        'fn': fn,
        'precision': precision,
        'recall': recall,
        'f1_score': f1,
        'accuracy': accuracy
    })

# Create DataFrame
threshold_df = pd.DataFrame(threshold_results)
threshold_df
```

```python
import plotly.graph_objects as go

# Create interactive plot with multiple metrics
fig = go.Figure()

# Add traces for each metric
fig.add_trace(go.Scatter(
    x=threshold_df['threshold'],
    y=threshold_df['precision'],
    mode='lines+markers',
    name='Precision',
    line=dict(color='blue', width=2),
    marker=dict(size=8)
))

fig.add_trace(go.Scatter(
    x=threshold_df['threshold'],
    y=threshold_df['recall'],
    mode='lines+markers',
    name='Recall',
    line=dict(color='green', width=2),
    marker=dict(size=8)
))

fig.add_trace(go.Scatter(
    x=threshold_df['threshold'],
    y=threshold_df['f1_score'],
    mode='lines+markers',
    name='F1 Score',
    line=dict(color='red', width=2),
    marker=dict(size=8)
))

fig.add_trace(go.Scatter(
    x=threshold_df['threshold'],
    y=threshold_df['accuracy'],
    mode='lines+markers',
    name='Accuracy',
    line=dict(color='purple', width=2),
    marker=dict(size=8)
))

# Add vertical line at current threshold (0.95)
fig.add_vline(
    x=0.95,
    line_dash="dash",
    line_color="gray",
    annotation_text="Current Threshold (0.95)",
    annotation_position="top"
)

# Update layout
fig.update_layout(
    title='Performance Metrics vs Threshold',
    xaxis_title='Threshold',
    yaxis_title='Score',
    hovermode='x unified',
    height=500,
    width=1000,
    legend=dict(
        yanchor="middle",
        y=0.5,
        xanchor="right",
        x=1.15,
        bgcolor="rgba(255, 255, 255, 0.8)",
        bordercolor="gray",
        borderwidth=1
    )
)

fig.show()

# Display metrics at current threshold (0.95)
print("\nMetrics at threshold = 0.95:")
print(threshold_df[threshold_df['threshold'] == 0.95].T)
```
