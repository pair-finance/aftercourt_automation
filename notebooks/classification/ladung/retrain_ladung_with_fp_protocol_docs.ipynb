preprocess fp data

```python
import os

notebook_dir = r'/Users/melih.gorgulu/Desktop/Projects/intent_recognition/notebooks'
project_dir = os.path.dirname(notebook_dir)
print("Notebook Directory:", notebook_dir)
print("Project Directory:", project_dir)
os.chdir(project_dir)
```

```python
from src.services.attachment_processing.base_input_processor import AfterCourtAttachmentPreprocessor
from src.domain.base.blueprints import AfterCourtPreprocessingBlueprint
```

```python
# Initialize preprocessor
config = AfterCourtPreprocessingBlueprint(
    normalize_whitespace=True,
    remove_short_lines=True,
    short_line_threshold=3,
    remove_html_tags=True,
    lowercase=True,
    clean_text_type='processed'
)

preprocessor = AfterCourtAttachmentPreprocessor(name='after_court', preprocessing_config=config)

# Apply replace_with_tags directly to each text

fp_data['text_w_tags'] = fp_data['texts'].apply(lambda x: preprocessor._replace_with_tags(preprocessor._process(x)))
```

```python
fp_data
```

```python
fp_data['document_type'] = 'fp_protocol'
fp_data['is_ladung'] = 0
```

```python
latest_full_data
```

```python
data = pd.concat([latest_full_data, fp_data])[['text_w_tags','document_type','is_ladung']]
data.reset_index(inplace=True,drop=True)
```

```python
data
```

```python
data['document_type'].value_counts()
```

get the vectorizer

```python
import dill
from sklearn.feature_extraction.text import TfidfVectorizer
# Make sure to import ClassificationSpacyLemmaTokenizer from the source code.
from src.services.models.aftercourt_tokenizer import ClassificationSpacyLemmaTokenizer

vec_path = "/Users/melih.gorgulu/Desktop/Projects/intent_recognition/notebooks/after-court/models/vectorizers/05-12-2026_tf_idf_vectorizer_v1.dill"
tf_idf_vectorizer = dill.load(open(vec_path, 'rb'))
```

train test split

```python

X_text = data['text_w_tags']
y = data['is_ladung']

# train test split
from sklearn.model_selection import train_test_split

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Train set size: {len(X_train_text)}, Test set size: {len(X_test_text)}")
print(f"Train set is pfub dist: {y_train.value_counts(normalize=True)}")
print(f"Test set is pfub dist: {y_test.value_counts(normalize=True)}")
```

```python
document_types_train = data.loc[X_train_text.index, 'document_type']
document_types_test = data.loc[X_test_text.index, 'document_type']
```

```python
document_types_train.value_counts()
```

```python
document_types_test.value_counts()
```

```python
X_train_tfidf = tf_idf_vectorizer.transform(X_train_text)
X_test_tfidf = tf_idf_vectorizer.transform(X_test_text)
```

```python
# basic RF

from sklearn.ensemble import RandomForestClassifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train_tfidf, y_train)
y_pred = rf_classifier.predict(X_test_tfidf)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

accuracy = accuracy_score(y_test, y_pred)


print(f"Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

"""
    true negatives is :math:`C_{0,0}`, false negatives is :math:`C_{1,0}`, 
    true positives is :math:`C_{1,1}` and false positives is :math:`C_{0,1}
"""
```

```python
import pandas as pd

# Create dataframe with predictions, actual values, and original indexes
results_df = pd.DataFrame({
    'index': y_test.index,
    'y_test': y_test.values,
    'y_pred': y_pred
})

results_df
```

```python
results_df['FP'] = (results_df['y_test'] == 0) & (results_df['y_pred'] == 1)
results_df['FN'] = (results_df['y_test'] == 1) & (results_df['y_pred'] == 0)
```
