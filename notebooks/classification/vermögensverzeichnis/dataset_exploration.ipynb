```python
data_new['created_at'].iloc[0]
```

```python
def create_object_key(row):
    created_at = row.created_at
    year, month, day = created_at.year, created_at.month, created_at.day
    
    if month < 10:
        month = f"0{month}"
    if day < 10:
        day = f"0{day}"
    
    date_folder = f"{day}.{month}.{year}/"
    object_key = date_folder + row.document_file_name
    return object_key
```

```python
data_new['object_key'] = data_new.apply(create_object_key, axis=1)
```

```python
# --- Comprehensive comparison of final_raw_data vs df_to_save ---

print("=" * 60)
print("SHAPE COMPARISON")
print("=" * 60)
print(f"final_raw_data: {final_raw_data.shape[0]} rows, {final_raw_data.shape[1]} cols")
print(f"df_to_save:     {df_to_save.shape[0]} rows, {df_to_save.shape[1]} cols")
print(f"New rows added:  {df_to_save.shape[0] - final_raw_data.shape[0]}")
new_cols = set(df_to_save.columns) - set(final_raw_data.columns)
removed_cols = set(final_raw_data.columns) - set(df_to_save.columns)
if new_cols:
    print(f"New columns:     {new_cols}")
if removed_cols:
    print(f"Removed columns: {removed_cols}")

print("\n" + "=" * 60)
print("DOCUMENT TYPE DISTRIBUTION COMPARISON")
print("=" * 60)
dist_old = final_raw_data['document_type'].value_counts()
dist_new = df_to_save['document_type'].value_counts()
dist_comparison = pd.DataFrame({'before': dist_old, 'after': dist_new}).fillna(0).astype(int)
dist_comparison['diff'] = dist_comparison['after'] - dist_comparison['before']
print(dist_comparison)

print("\n" + "=" * 60)
print("NEW ROWS DETAIL")
print("=" * 60)
if 'object_key' in df_to_save.columns:
    old_keys = set(final_raw_data['object_key'].dropna())
    new_keys = set(df_to_save['object_key'].dropna())
    added_keys = new_keys - old_keys
    print(f"New object_keys added: {len(added_keys)}")
    new_rows = df_to_save[df_to_save['object_key'].isin(added_keys)]
    print(f"\nNew rows breakdown by document_type:")
    print(new_rows['document_type'].value_counts())
    if 'is_ve_with_invoice' in new_rows.columns:
        print(f"\nNew rows breakdown by is_ve_with_invoice:")
        print(new_rows['is_ve_with_invoice'].value_counts())
    print(f"\nSample of newly added rows:")
    display(new_rows.head(10))
else:
    new_rows = df_to_save.iloc[final_raw_data.shape[0]:]
    print(f"New rows (by index): {len(new_rows)}")
    display(new_rows.head(10))

print("\n" + "=" * 60)
print("NULL VALUE COMPARISON")
print("=" * 60)
nulls_old = final_raw_data.isnull().sum()
nulls_new = df_to_save.isnull().sum()
common_cols = sorted(set(final_raw_data.columns) & set(df_to_save.columns))
null_comparison = pd.DataFrame({
    'before_nulls': nulls_old.reindex(common_cols),
    'after_nulls': nulls_new.reindex(common_cols)
}).fillna(0).astype(int)
null_comparison['diff'] = null_comparison['after_nulls'] - null_comparison['before_nulls']
null_comparison = null_comparison[null_comparison['diff'] != 0]
if not null_comparison.empty:
    print(null_comparison)
else:
    print("No change in null counts across shared columns.")

print("\n" + "=" * 60)
print("DUPLICATE CHECK")
print("=" * 60)
if 'object_key' in df_to_save.columns:
    dupes = df_to_save[df_to_save['object_key'].duplicated(keep=False) & df_to_save['object_key'].notna()]
    print(f"Duplicate object_keys in df_to_save: {dupes['object_key'].nunique()} keys ({len(dupes)} rows)")
    if not dupes.empty:
        display(dupes.head(10))
```
