## Drittauskunft Heuristics



### 1. Source / issuer

- The document is issued by the **Bundeszentralamt für Steuern (BZSt)**.

- English: **Federal Central Tax Office**.

- Interpretation: this is a strong contextual signal, but not sufficient on its own.



### 2. Cases with no returned result should still count as drittauskunft

These phrases indicate that the lookup was attempted and belongs to the drittauskunft process, even when no usable result was returned.



- **Der Abruf hat zu keinem Ergebnis geführt.**

  English: **The retrieval returned no result.**



- **Die Anfrage lieferte kein verwertbares Ergebnis.**

  English: **The request did not yield any usable result.**



- **Das BZSt teilt mit, dass der Abruf zu keinem Ergebnis geführt hat.**

  English: **The BZSt states that the retrieval returned no result.**



### 3. Strong indicators for drittauskunft



#### High importance

- **in o. g. Sache haben Sie mich beauftragt, eine Drittauskunft nach § 802l ZPO beim Bundeszentralamt für Steuern einzuholen**

  English: **In the above-mentioned matter, you instructed me to obtain third-party information pursuant to Section 802l ZPO from the Federal Central Tax Office.**

  Interpretation: explicit statement of the drittauskunft process.



- **Das Ergebnis teile ich Ihnen unter Übersendung einer Kopie der entsprechenden Stellen mit.**

  English: **I am informing you of the result and enclosing a copy of the relevant passages.**

  Interpretation: typical wording when communicating retrieved third-party information.



- **… erhalten Sie … in der Anlage das Ergebnis der Auskunft beim Bundeszentralamt für Steuern zur Kenntnis.**

  English: **... enclosed please find the result of the information request from the Federal Central Tax Office for your information.**

  Interpretation: explicit reference to transmitted drittauskunft results.



- **Kontenabrufersuchen nach §§ 93, 93b Abgabenordnung (AO)**

  English: **Account retrieval request pursuant to Sections 93, 93b of the Fiscal Code (AO).**

  Interpretation: very strong domain-specific indicator tied to official data retrieval.



- **Ergebnis / Ergebnisse**

  English: **Result / Results**

  Interpretation: relevant only in combination with other BZSt or drittauskunft markers; not sufficient alone.



#### Medium importance

- **Bundeszentralamt für Steuern / BZSt**

  English: **Federal Central Tax Office / BZSt**

  Interpretation: useful institutional marker, but too broad as a standalone feature.



- **St II 4 - S 0229a - KEVIZZ**

  English: likely an **internal BZSt reference / file number / department code**.

  Short interpretation: probably a routing or classification code used in official BZSt correspondence; useful as a pattern marker when it appears with other drittauskunft language.



- **Anlage zu GZ: St II 4 - S 0229a - KEVIZZ - …**

  English: **Attachment to ref. no.: St II 4 - S 0229a - KEVIZZ - ...**

  Short interpretation: likely an attachment tied to a BZSt file or transaction reference.



- **…die Anfrage lieferte kein verwertbares Ergebnis. Das BZSt teilt mit, dass … zu keinem Ergebnis geführt hat.**

  English: **...the request did not yield any usable result. The BZSt states that ... returned no result.**

  Interpretation: medium strength because it reflects the correct process and issuer, even without a positive hit.



- **Die von Ihnen beantragten Drittstellenauskünfte liegen an.**

  English: **The third-party information you requested is enclosed.**

  Interpretation: strong phrasing, but slightly less explicit than the top signals above.



#### Low importance

- **Der Abruf hat zu keinem Ergebnis geführt.**

  English: **The retrieval returned no result.**

  Interpretation: useful only with surrounding BZSt / drittauskunft context.



- **Nach Zweckerreichung sind erlangte Schuldnerdaten zu löschen.**

  English: **Once the purpose has been fulfilled, the debtor data obtained must be deleted.**

  Interpretation: compliance / retention language that may appear in this document family, but is weak on its own.


```python
import re

# OCR-robust regex helpers
# - Whitespace (including newlines, NBSP, OCR-inserted spaces) is allowed between tokens.
# - German umlauts are matched with tolerant alternatives (ä -> (?:ä|ae|a),
#   ö -> (?:ö|oe|o), ü -> (?:ü|ue|u), ß -> (?:ß|ss|s)), which is common OCR confusion.
# - "<anything_here>" / "<something>" placeholders become a non-greedy wildcard
#   that can span a small amount of text including line breaks.
# - All patterns are compiled with re.IGNORECASE | re.DOTALL | re.UNICODE.

WS = r"[\s\u00A0]+"                # any run of whitespace incl. NBSP
GAP = r"[\s\S]{0,80}?"             # tolerant non-greedy gap for <placeholder>

def _tok(token: str) -> str:
    """Escape a single word/token and make umlauts OCR-tolerant."""
    out = []
    for ch in token:
        if ch == "ä":
            out.append(r"(?:ä|ae|a)")
        elif ch == "ö":
            out.append(r"(?:ö|oe|o)")
        elif ch == "ü":
            out.append(r"(?:ü|ue|u)")
        elif ch == "Ä":
            out.append(r"(?:Ä|Ae|A)")
        elif ch == "Ö":
            out.append(r"(?:Ö|Oe|O)")
        elif ch == "Ü":
            out.append(r"(?:Ü|Ue|U)")
        elif ch == "ß":
            out.append(r"(?:ß|ss|s)")
        else:
            out.append(re.escape(ch))
    return "".join(out)

def build_pattern(phrase: str) -> str:
    """Convert a phrase into an OCR-robust regex.

    - splits on whitespace and joins tokens with a flexible whitespace matcher
    - replaces <...> placeholders with a tolerant gap
    """
    parts = []
    for token in phrase.split():
        m = re.fullmatch(r"<[^>]+>", token)
        if m:
            parts.append(GAP)
        else:
            parts.append(_tok(token))
    # join with flexible whitespace; placeholders already allow whitespace inside
    pattern = ""
    for i, p in enumerate(parts):
        if i == 0:
            pattern = p
        else:
            if p == GAP or parts[i - 1] == GAP:
                pattern += p
            else:
                pattern += WS + p
    return pattern


high_importance_features = [
    "Sache haben Sie mich beauftragt, eine Drittauskunft nach <anything_here> ZPO beim Bundeszentralamt für Steuern einzuholen",
    "Das Ergebnis teile ich Ihnen unter Übersendung einer Kopie der entsprechenden Stellen mit",
    "in der Anlage das Ergebnis der Auskunft beim Bundeszentralamt für Steuern zur Kenntnis",
    "Kontenabrufersuchen nach <something> Abgabenordnung",
    "Ergebnis",
    "Ergebnisse",
]

FLAGS = re.IGNORECASE | re.DOTALL | re.UNICODE

high_importance_patterns = [
    re.compile(build_pattern(phrase), FLAGS) for phrase in high_importance_features
]

for phrase, pat in zip(high_importance_features, high_importance_patterns):
    print(f"\n# {phrase}\n{pat.pattern}")
```

```python
high_importance_idx_to_pattern = {i: pat for i, pat in enumerate(high_importance_patterns)}
```

```python
pattern_found_column = []
for i, text in enumerate(data.text):
    found_patterns = []
    for j, pat in high_importance_idx_to_pattern.items():
        if pat.search(text):
            found_patterns.append(j)
    pattern_found_column.append(found_patterns)
    
data["high_importance_patterns_found"] = pattern_found_column
    
```

```python
no_patterns_found = data[data.high_importance_patterns_found.apply(len) == 0]
no_patterns_found.document_type.value_counts()
```

```python
pattern_found = data[data.high_importance_patterns_found.apply(len) > 0]
pattern_found.document_type.value_counts()
```

```python
pattern_found['number_of_high_importance_patterns'] = pattern_found.high_importance_patterns_found.apply(len)
```

```python
# plot histogram for each group
import matplotlib.pyplot as plt
for doc_type, group in pattern_found.groupby("document_type"):
    plt.figure()
    group.number_of_high_importance_patterns.hist(bins=range(0, group.number_of_high_importance_patterns.max() + 2))
    plt.title(f"Document Type: {doc_type}")
    plt.xlabel("Number of High Importance Patterns Found")
    plt.ylabel("Frequency")
    plt.xticks(range(0, group.number_of_high_importance_patterns.max() + 1))
    plt.grid(axis="y")
```
