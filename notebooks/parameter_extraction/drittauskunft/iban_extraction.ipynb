```python

import re
NO_RESULT_PATTERN_1 = re.compile(
    r"\bkein(?:em|en|er|es|e)?\b"  # kein / keinem / keinen / keiner / keines / keine
    r"(?:\s+\w+){0,4}?"  # up to 4 optional words in between (e.g. "verwertbares")
    r"\s+ergebniss?(?:e|en|es)?\b",  # ergebnis / ergebnisse / ergebnissen / ergebnisses
    flags=re.IGNORECASE,
)

NO_RESULT_PATTERN_2 = re.compile(
    r"der\s+abruf\s+hat\s+zu\s+keinem\s+ergebnis\s+gef(?:ü|ue)hrt",
    re.IGNORECASE,
)

DRITTAUSKUNFT_BAK_NUMBER_REGEX = re.compile(r"\bbak\s*-\s*(?:nr|nummer)\.?:?\s*\d{6}\b", re.IGNORECASE)


def extract_no_result(text: str) -> bool:
    """
    Check if the document contains phrases indicating no results.

    Args:
        text (str): Input text to check for "kein Ergebnis" phrases

    Returns:
        bool: True if "kein Ergebnis" is found, False otherwise
    """
    if not text:
        return False

    # Search for the "kein Ergebnis" pattern in the text
    match = NO_RESULT_PATTERN_1.search(text) or NO_RESULT_PATTERN_2.search(text)
    return match is not None


def extract_bank_accounts(text: str) -> list:
    """Extract unique 6 digit BAK Numbers from results

    Args:
        text (str): OCR parsed text, detected as drittauskunft

    Returns:
        List of extracted Unique BAK Numbers
    """
    if not text:
        return []
    # search for BAK-NR pattern.
    matches = DRITTAUSKUNFT_BAK_NUMBER_REGEX.findall(text)
    unique_bak_numbers = set(re.search(r"\d{6}", match).group() for match in matches if re.search(r"\d{6}", match))
    return list(unique_bak_numbers)
```

```python

GENERAL_IBAN_REGEX = re.compile(r'\b[A-Z]{2}\d{2}(?:\s?[A-Z0-9]{4}){2,7}\s?[A-Z0-9]{1,3}\b')

def validate_iban(iban: str) -> bool:
    """
    Validate an IBAN using the standard IBAN validation algorithm.
    
    Referance: https://en.wikipedia.org/wiki/International_Bank_Account_Number

    Args:
        iban (str): The IBAN to validate.

    Returns:
        bool: True if the IBAN is valid, False otherwise.
    """
    # Remove spaces and convert to uppercase
    iban = iban.replace(' ', '').upper()
    
    # Check if the IBAN length is valid (between 15 and 34 characters)
    if not (15 <= len(iban) <= 34):
        return False
    
    # Move the first four characters to the end of the string
    rearranged_iban = iban[4:] + iban[:4]
    
    # Replace each letter in the string with two digits, where A=10, B=11, ..., Z=35
    numeric_iban = ''
    for char in rearranged_iban:
        if char.isdigit():
            numeric_iban += char
        else:
            numeric_iban += str(ord(char) - 55)  # Convert letter to number (A=10, B=11, ..., Z=35)
    
    # Convert the string to an integer and compute the remainder of that number on division by 97
    iban_int = int(numeric_iban)
    
    return iban_int % 97 == 1


def _normalize_iban(iban: str) -> str:
    """Canonical form used for uniqueness checks (no spaces, uppercase)."""
    return iban.replace(" ", "").upper()



DATE_REGEX = re.compile(
    r"""(?x)
(?<!\d)
(?:
    (?:(?:0?[1-9]|[12]\d)|3[01])\s?[./:-][\s.]?(?:0?[13578]|1[02]|J(?:an(?:uar)?|uli?)|M(?:ärz?|ai)|Aug(?:ust)?|Dez(?:ember)?|Okt(?:ober)?)\s?(?:[./:-][\s.]?)?(?:[1-9]\d\d\d|\d\d)|
    (?:(?:0?[1-9]|[12]\d)|30)\s?[./:-][\s.]?(?:0?[13-9]|1[012]|J(?:an(?:uar)?|u[nl]i?)|M(?:ärz?|ai)|A(?:pr(?:il)?|ug(?:ust)?)|Sep(?:tember)?|(?:Nov|Dez)(?:ember)?|Okt(?:ober)?)\s?(?:[./:-][\s.]?)?(?:[1-9]\d\d\d|\d\d)|
    (?:0?[1-9]|[12]\d)\s?[./:-][\s.]?(?:0?2|Fe(?:b(?:ruar)?)?|(?:0?2|Feb(?:r(?:uar)?)?))\s?(?:[./:-][\s.]?)?(?:[1-9]\d(?:[02468][048]|[13579][26])|\d\d)|
    (?:0?[1-9]|[12][0-9])\s?[./:-][\s.]?(?:0?2|Fe(?:b(?:ruar)?)?|(?:0?2|Feb(?:r(?:uar)?)?))\s?(?:[./:-][\s.]?)?(?:[1-9]\d\d\d|\d\d)
)
(?!\d)""",
    re.IGNORECASE | re.VERBOSE,
)

import dateparser  # same parser used by the Ladung summon_date_extractor
from datetime import datetime


def extract_date_from_context(context: str) -> datetime | None:
    """Extract the Errichtungsdatum from an IBAN subcontext.

    Reuses the DATE_REGEX and dateparser from the Ladung summon_date_extractor.
    The subcontext spans from an IBAN to the next Kontoinhaber, so it typically
    contains a single relevant date. The first parseable date found is returned.

    Args:
        context (str): Text slice starting at an IBAN and ending before the next
            Kontoinhaber (or end of context).

    Returns:
        datetime | None: The parsed date, or None if no valid date is found.
    """
    if not context:
        return None

    for match in DATE_REGEX.finditer(context):
        parsed = dateparser.parse(match.group(), languages=["de"])
        if parsed is not None:
            return parsed

    return None

def get_latest_iban(ibans: list[str], ibans_date: list[datetime | None]) -> str | None:
    """Return the IBAN with the most recent Errichtungsdatum.

    Args:
        ibans (list[str]): Normalized IBANs found for a single BAK context.
        ibans_date (list[datetime | None]): The Errichtungsdatum for each IBAN,
            aligned by index with ``ibans``. May contain None when no date was
            found.

    Returns:
        str | None: The IBAN with the latest date. If none of the IBANs have a
            date, falls back to the last IBAN found. Returns None if there are
            no IBANs.
    """
    if not ibans:
        return None

    dated_ibans = [(iban, date) for iban, date in zip(ibans, ibans_date) if date is not None]
    if dated_ibans:
        return max(dated_ibans, key=lambda pair: pair[1])[0]

    # No dates available -> fall back to the last IBAN found.
    return ibans[-1]

def extract_iban_from_contexts(contexts: list[str], bak_numbers: list[str]) -> dict[str, str]:
    # TODO: New task: we also need to extract errictungsdatum of the iban
    bak_to_iban = {}
    seen_ibans = set()  # normalized IBANs already assigned to a BAK
    for ctx, bak in zip(contexts, bak_numbers):
        extracted_ibans = []
        ibans_date = []
        for iban in GENERAL_IBAN_REGEX.findall(ctx):
            if not validate_iban(iban):
                continue
            norm = _normalize_iban(iban)
            if norm in seen_ibans:
                continue
            # iban is found for the current bak number context. Now we need to extract its date.
            # for that search in subcontext to prevent fetching multiple dates
            # subcontext = starts with iban and ends with first next Kontoinhaber
            subcontext_start = ctx.find(iban) # start index of the iban in the context
            first_kontinhaber_match = next(KONTOINHABER_REGEX.finditer(ctx[subcontext_start:]), None)
            if first_kontinhaber_match:
                subcontext_end = subcontext_start + first_kontinhaber_match.start()
            else:
                # if no Kontoinhaber is found, the end is  nd of the context
                subcontext_end = len(ctx)
            
            extracted_date = extract_date_from_context(ctx[subcontext_start:subcontext_end])
            # add iban
            seen_ibans.add(norm)
            extracted_ibans.append(norm)
            ibans_date.append(extracted_date)
            
        latest_iban = get_latest_iban(extracted_ibans, ibans_date)
            
        bak_to_iban[bak] = latest_iban
    return bak_to_iban
```

```python
def extract_bak_to_ibans(text: str) -> dict[str, list[str]]:
    """Extract a mapping of BAK numbers to their validated IBANs from a Drittauskunft document.

    Pipeline:
        1. Find all BAK-Nr. matches in the text.
        2. Deduplicate BAK numbers (keep first occurrence).
        3. Split the text into a context window per BAK.
        4. Extract and validate IBANs within each context.

    Args:
        text (str): OCR parsed text of a document detected as drittauskunft.

    Returns:
        dict[str, list[str]]: Mapping of BAK number -> list of validated IBANs.
    """
    if not text:
        return {}

    match_objects = list(DRITTAUSKUNFT_BAK_NUMBER_REGEX.finditer(text))
    filtered_match_objects = filter_bak_spans(match_objects)
    contexts, bak_numbers = split_to_contexts(filtered_match_objects, text)
    return extract_iban_from_contexts(contexts, bak_numbers)
```
